## 3. Methodology

This project uses a supervised regression framework to study which institutional characteristics are associated with six-year bachelor's graduation rates. Since the outcome variable is a numerical percentage, the main method is multiple linear regression rather than classification. The goal is not only to predict graduation rates, but also to interpret how admissions, cost, financial aid, and institution type are related to graduation outcomes after controlling for other variables.

The response variable is the total six-year bachelor's graduation rate, reported in IPEDS as `DRVGR2023_RV.Graduation rate - Bachelor degree within 6 years, total`. This measures the percentage of students in the bachelor's cohort who completed a bachelor's degree within six years. We use the six-year rate because it is a common completion measure and gives a more complete view of student success than only using the four-year graduation rate.

The explanatory variables are selected from the main categories in the dataset. Admissions variables, including percent admitted and admissions yield, are used to represent selectivity and enrollment behavior. Cost variables include tuition and fees for 2023-24 and total price measures. Institutional variables include control of institution, sector, size category, and urbanization level. Financial aid variables, such as the percent of students receiving aid and average aid amounts, are also considered because affordability may be related to persistence and completion.

The general multiple linear regression model is written as

$$GraduationRate_i = \beta_0 + \beta_1Yield_i + \beta_2Tuition_i + \beta_3Price_i + \beta_4InstitutionType_i + \cdots + \varepsilon_i.$$

Here, each institution is one observation. The intercept \(\beta_0\) represents the expected graduation rate for the reference group when the numerical predictors are equal to zero, while each slope coefficient represents the expected change in graduation rate, measured in percentage points, holding the other predictors constant. The error term \(\varepsilon_i\) represents variation in graduation rates that is not explained by the predictors in the model.

Categorical predictors cannot be entered into the regression model directly, so they are converted into dummy variables. For example, institution control can be represented by indicator variables comparing public and private nonprofit institutions, with one category used as the reference group. This allows the model to estimate average differences in graduation rates between institution types while also adjusting for tuition, admissions, and other numerical predictors.

The model fitting procedure begins with a baseline regression containing the main predictors listed in the project outline: admissions yield, tuition and fees, total price, and institution type. After this baseline model is fit, additional variables such as institution size, urbanization level, percent admitted, and selected financial aid measures are added to form expanded models. This step-by-step approach makes the modeling process easier to interpret because we can see whether the added variables meaningfully improve the model.

Model comparison is based on both fit and interpretability. We compare models using R-squared and adjusted R-squared, where adjusted R-squared is especially useful because it accounts for the number of predictors in the model. We also consider residual standard error to understand the typical size of prediction errors. A larger model is not automatically preferred; the final model should explain graduation rates reasonably well while still being clear and connected to the research question.

Because several predictors may measure related concepts, multicollinearity is also checked. For example, tuition and total price are both cost-related variables, so they may be strongly correlated. If predictors are highly correlated, the model coefficients can become unstable and harder to interpret. In that case, we compare alternative models and keep the version that gives the clearest explanation of the relationships in the data.

After fitting the model, diagnostic checks are used to evaluate whether the regression assumptions are reasonable. Residual plots are used to check linearity and constant variance. A normal Q-Q plot is used to examine whether the residuals are approximately normal. Influential observations are checked using leverage and Cook's distance, since a small number of unusual institutions could have a large effect on the fitted model.

The final interpretation focuses on both statistical significance and practical meaning. P-values help identify which predictors show evidence of an association with graduation rates, while coefficient sizes and signs show the direction and size of each relationship. Since the data are observational, the regression model cannot prove causation. Instead, it is used to identify which institutional characteristics are most strongly associated with graduation rates after accounting for the other variables included in the model.

## 5. Results

This section examines institutional characteristics and their impact on graduation rates. We focus on students earning a bachelor's degree within six years.
To analyze this, we use multiple linear regression to build three models. We begin with a baseline model. Then We add more information to improve its explanatory power.We compare these models based on their goodness-of-fit. Specifically, we evaluate the adjusted $R^2$ and residual standard error. Finally, we use diagnostic plots to check the core assumptions of regression.

```python
# Convenience OLS helper: returns (coeffs, SE, t, p-vals, R2, adj-R2, RSE, fitted)
def fit_ols(X_arr, y):
    Xm = np.column_stack([np.ones(len(X_arr)), X_arr])
    n, p = Xm.shape
    coeffs, _, _, _ = np.linalg.lstsq(Xm, y, rcond=None)
    y_pred  = Xm @ coeffs
    ss_res  = np.sum((y - y_pred) ** 2)
    ss_tot  = np.sum((y - y.mean()) ** 2)
    r2      = 1 - ss_res / ss_tot
    adj_r2  = 1 - (1 - r2) * (n - 1) / (n - p - 1)
    rse     = np.sqrt(ss_res / (n - p))
    XtXinv  = np.linalg.pinv(Xm.T @ Xm)
    se      = np.sqrt(np.diag(ss_res / (n - p) * XtXinv))
    t_stat  = coeffs / se
    pvals   = 2 * (1 - stats.t.cdf(np.abs(t_stat), df=n - p))
    return coeffs, se, t_stat, pvals, r2, adj_r2, rse, y_pred

# Encode size dummies (reference category = '1,000 - 4,999')
size_dummies = pd.get_dummies(analysis_data['size'], drop_first=False).astype(int)
size_dummies = size_dummies.drop(columns=['1,000 - 4,999'], errors='ignore')
size_cols    = list(size_dummies.columns)

adr = pd.concat([analysis_data.copy(), size_dummies], axis=1)
adr['is_private'] = (adr['control'] == 'Private not-for-profit').astype(int)
adr['price_k']    = adr['price_in_state_on_campus'] / 1000   # scale to $thousands
adr['grant_k']    = adr['grant_amount'] / 1000

y = adr['grad6'].values
print(f'Results sample: n = {len(y)} institutions')
```

### 5.1 Baseline model

```python
# Model 1 — Baseline
# Predictors: admissions yield, total in-state price (per $1,000), institution type
m1_vars  = ['yield_rate', 'price_k', 'is_private']
X1       = adr[m1_vars].values
c1, se1, t1, p1, r2_1, ar2_1, rse1, yhat1 = fit_ols(X1, y)

names1   = ['Intercept'] + m1_vars
sig      = lambda p: '***' if p < 0.001 else '**' if p < 0.01 else '*' if p < 0.05 else ''

m1_table = pd.DataFrame({
    'Coefficient': c1.round(4),
    'Std. Error':  se1.round(4),
    't-value':     t1.round(3),
    'p-value':     [f'{v:.4f}' for v in p1],
    'Sig.':        [sig(v) for v in p1]
}, index=names1)

print(f'Model 1 — Baseline')
print(f'R\u00b2 = {r2_1:.4f}   Adj. R\u00b2 = {ar2_1:.4f}   RSE = {rse1:.3f}   n = {len(y)}')
print()
print(m1_table.to_string())
```

The baseline model includes admissions yield, total in-state price (scaled to thousands of dollars), and institution type (Private not-for-profit = 1; Public = 0). Together these three predictors explain 47.5% of the variance in six-year graduation rates (adj. R² = 0.473, RSE = 13.4 percentage points), and all coefficients are statistically significant (p < 0.001).

Each additional thousand dollars in total price is associated with a 0.89 percentage-point increase in graduation rate, holding other variables constant. The large negative coefficient on `is_private` (−20.5 pp) may seem counterintuitive, but reflects that price is held constant: among institutions charging similar amounts, public schools graduate more students. Because private institutions charge much higher prices, controlling for price reverses the raw private premium seen in the EDA. This signals that price alone does not capture all sector differences, and that additional predictors are needed.

### 5.2 Extended model

```python
# Model 2 — Add admissions rate and Pell Grant percentage
m2_vars  = ['yield_rate', 'price_k', 'admit_rate', 'pell_pct', 'is_private']
X2       = adr[m2_vars].values
c2, se2, t2, p2, r2_2, ar2_2, rse2, yhat2 = fit_ols(X2, y)

names2   = ['Intercept'] + m2_vars
m2_table = pd.DataFrame({
    'Coefficient': c2.round(4),
    'Std. Error':  se2.round(4),
    't-value':     t2.round(3),
    'p-value':     [f'{v:.4f}' for v in p2],
    'Sig.':        [sig(v) for v in p2]
}, index=names2)

print('Model 2 — Add admit_rate and pell_pct')
print(f'R\u00b2 = {r2_2:.4f}   Adj. R\u00b2 = {ar2_2:.4f}   RSE = {rse2:.3f}   n = {len(y)}')
print()
print(m2_table.to_string())
```

Adding admissions rate and Pell Grant percentage raises the adjusted R² from 0.473 to 0.599 (a 13 percentage points increase) and reduces the RSE from 13.4 to 11.7 pp. This confirms that selectivity and student income composition are important predictors of graduation outcomes beyond price and sector alone.

**Admissions rate** (`admit_rate`): each one-percentage-point increase in the admission rate (i.e., a less selective institution) is associated with a 0.15 pp decrease in graduation rate (p < 0.001), holding other variables constant. More selective institutions enroll students with stronger academic preparation, which is reflected in higher completion rates.

**Pell Grant share** (`pell_pct`): each one-percentage-point increase in the share of students receiving Pell Grants is associated with a 0.42 pp decrease in graduation rate (p < 0.001). Since Pell eligibility reflects lower household income, this is among the strongest signals in the model: financial barriers and socioeconomic disadvantage remain strongly associated with lower completion rates even after controlling for tuition costs.

After controlling for these two variables, the private sector penalty shrinks from −20.5 to −13.2 pp, confirming that part — but not all — of the baseline gap was explained by omitted variables.

### 5.3 Multicollinearity check

```python
# Multicollinearity check — VIF for Model 2
# Note: tuition and price_in_state_on_campus have r = 0.985, so only price_k is retained.
Xm2 = np.column_stack([np.ones(len(X2)), X2])
vif_rows = []
for i, nm in enumerate(names2):
    if nm == 'Intercept':
        continue
    idx_rest = [j for j in range(Xm2.shape[1]) if j != i]
    xi       = Xm2[:, i]
    xrest    = Xm2[:, idx_rest]
    c_, _, _, _ = np.linalg.lstsq(xrest, xi, rcond=None)
    xhat  = xrest @ c_
    ss_r  = np.sum((xi - xhat) ** 2)
    ss_t  = np.sum((xi - xi.mean()) ** 2)
    r2i   = 1 - ss_r / ss_t
    vif_rows.append({'Variable': nm, 'VIF': round(1 / (1 - r2i), 2)})

print('Variance Inflation Factors — Model 2')
print(pd.DataFrame(vif_rows).to_string(index=False))
print()
print('Rule of thumb: VIF < 5 = no serious concern, VIF > 10 = problematic.')
```

`tuition` and `price_in_state_on_campus` are correlated at r = 0.985, so only `price_k` (total in-state price) is kept to avoid near-perfect multicollinearity. The VIF values for all five remaining predictors in Model 2 are below 4, well within the commonly used threshold of 5–10. There is therefore no serious multicollinearity concern, and the estimated coefficients can be interpreted reliably.

### 5.4 Full model

```python
# Model 3 — Full model (add grant aid, loan percentage, institution size)
m3_vars  = ['yield_rate', 'price_k', 'admit_rate', 'pell_pct',
            'grant_k', 'loan_pct', 'is_private'] + size_cols
X3       = adr[m3_vars].values
c3, se3, t3, p3, r2_3, ar2_3, rse3, yhat3 = fit_ols(X3, y)

names3   = ['Intercept'] + m3_vars
m3_table = pd.DataFrame({
    'Coefficient': c3.round(4),
    'Std. Error':  se3.round(4),
    't-value':     t3.round(3),
    'p-value':     [f'{v:.4f}' for v in p3],
    'Sig.':        [sig(v) for v in p3]
}, index=names3)

print('Model 3 — Full model')
print(f'R\u00b2 = {r2_3:.4f}   Adj. R\u00b2 = {ar2_3:.4f}   RSE = {rse3:.3f}   n = {len(y)}')
print()
print(m3_table.to_string())
```

Model 3 adds average grant aid, loan percentage, and institution size dummies to Model 2. The adjusted R² increases from 0.599 to 0.652, and the RSE drops to 10.9 percentage points. 
Institution size shows a clear pattern: larger institutions (10,000+ students) have significantly higher graduation rates than the reference group (1,000–4,999), while the smallest institutions (under 1,000) graduate fewer students on average. 
However, `yield_rate` becomes non-significant in this model (p = 0.176), suggesting it no longer adds explanatory value once size and grant aid are controlled for.

### 5.5 Model comparison

Model 3 achieves the best fit by all measures. However, two considerations favour Model 2 as the preferred model for interpretation.

Second, Model 2 retains all the key research variables and already explains 60% of the variance with only five predictors, making it substantially easier to interpret. 
**Model 2 is therefore selected as the final model.**

### 5.6 Model diagnostics

```python
# Diagnostic plots — Model 2 (final model)
resid2    = y - yhat2
std_resid = resid2 / rse2

Xm2      = np.column_stack([np.ones(len(X2)), X2])
n_obs, p_hat = Xm2.shape
H        = Xm2 @ np.linalg.pinv(Xm2.T @ Xm2) @ Xm2.T
leverage = np.diag(H)
cooks_d  = (resid2 ** 2 / (p_hat * rse2 ** 2)) * (leverage / (1 - leverage) ** 2)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 1. Residuals vs Fitted
axes[0].scatter(yhat2, resid2, alpha=0.3, s=10, color='steelblue')
axes[0].axhline(0, color='red', linewidth=1, linestyle='--')
axes[0].set_xlabel('Fitted values')
axes[0].set_ylabel('Residuals')
axes[0].set_title('Residuals vs Fitted')

# 2. Normal Q-Q
(osm, osr), (slope, intercept, _) = stats.probplot(std_resid, dist='norm')
axes[1].scatter(osm, osr, alpha=0.3, s=10, color='steelblue')
axes[1].plot(osm, slope * np.array(osm) + intercept, color='red', linewidth=1)
axes[1].set_xlabel('Theoretical quantiles')
axes[1].set_ylabel('Standardised residuals')
axes[1].set_title('Normal Q-Q Plot')

# 3. Cook's distance
axes[2].stem(range(len(cooks_d)), cooks_d, markerfmt=',',
             linefmt='steelblue', basefmt='grey')
axes[2].axhline(0.5, color='red', linestyle='--', linewidth=1, label="Cook's D = 0.5")
axes[2].set_xlabel('Observation index')
axes[2].set_ylabel("Cook's distance")
axes[2].set_title("Cook's Distance")
axes[2].legend(fontsize=9)

plt.tight_layout()
plt.show()

print(f"High-leverage points (h > 2p/n = {2*p_hat/n_obs:.3f}): {(leverage > 2*p_hat/n_obs).sum()}")
print(f"Cook's D > 0.5: {(cooks_d > 0.5).sum()}")
```

**Residuals vs Fitted.** The residuals are roughly symmetrically scattered around zero across the range of fitted values with no obvious curvature, supporting the assumptions of linearity and approximately constant variance. There is a slight suggestion of marginally larger spread at intermediate fitted values, but not severe enough to invalidate the model.

**Normal Q-Q plot.** The standardised residuals follow the reference line closely in the centre of the distribution. The tails show minor departure (a small number of institutions with unusually low or high graduation rates relative to model predictions). This mild tail behaviour is common in real-world data and does not substantially affect inference given the large sample size (n = 1,423).

**Cook's distance.** No observation has a Cook's distance above 0.5, so there are no unduly influential institutions. Although 126 observations have high leverage (h > 2p/n), none pair high leverage with a large residual, meaning the fitted coefficients are stable with respect to individual data points.

Overall, the diagnostics support the use of Model 2 for inference. The slight heteroscedasticity and non-normality in the tails are worth noting when interpreting borderline p-values, but they do not undermine the main conclusions.

## 6. Conclusion

This project examined which institutional factors are associated with six-year bachelor's graduation rates among four-year degree-granting public and private non-profit institutions in the United States. Using IPEDS 2023 data, the final analysis sample contained 1,423 institutions after restricting the sample and removing observations with missing values. The response variable was the six-year bachelor's graduation rate, and the main predictors included admissions selectivity, total price, Pell Grant percentage, institutional control, grant aid, loan percentage, and institution size.

Overall, the regression results suggest that graduation rates are strongly associated with both academic and financial characteristics of institutions. In the preferred Model 2, higher total price was positively associated with graduation rate, while higher admission rate and higher Pell Grant percentage were negatively associated with graduation rate. This suggests that more selective institutions and institutions serving fewer low-income students tend to have higher graduation rates, even after controlling for other variables. Private non-profit institutions also showed lower predicted graduation rates than public institutions after controlling for price, admissions, and Pell Grant percentage, although this result should be interpreted carefully because institution type is related to many other institutional characteristics.

The full model improved the adjusted R-squared from 0.599 to 0.652 and lowered the residual standard error from 11.656 to 10.857, showing that adding financial aid, loan percentage, and size variables improved predictive performance. However, Model 2 was selected as the preferred model because it was easier to interpret while still explaining about 60% of the variation in graduation rates. The diagnostics also suggested that the model was reasonably reliable: residuals were centered around zero, there were no highly influential observations based on Cook's distance, and the main conclusions were not strongly affected by outliers.

There are several limitations to this study. First, the analysis is observational, so the results show association rather than causation. For example, higher price may not directly cause higher graduation rates; instead, it may reflect institutional resources, selectivity, reputation, or student background. Second, the complete-case approach removed some observations with missing data, which may affect the representativeness of the final sample. Third, the model does not include all possible factors that influence graduation outcomes, such as student test scores, family income, retention programs, faculty resources, or campus-level academic support.

Future research could extend this project by using multiple years of IPEDS data to study trends over time. It would also be useful to compare different modeling approaches, such as regularized regression, random forests, or classification models for high- versus low-graduation-rate institutions. Another possible extension would be to include more detailed student and institutional variables, such as retention rate, student-faculty ratio, endowment, or regional economic factors. These additions could help explain more variation in graduation rates and provide a deeper understanding of which institutional factors are most closely related to student success.
